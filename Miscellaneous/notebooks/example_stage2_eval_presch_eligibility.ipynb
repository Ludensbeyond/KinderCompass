```python
import datetime

def evaluate_preschool_eligibility(dob, admission_date, ghi, base_fee, basic_subsidy):
    """
    KinderCompass Stage 2: Decision Automation Engine ⚙️
    Computes age compliance, additional subsidy brackets, and final net fees 
    based on official Early Childhood Development Agency (ECDA) guidelines.
    """
    
    # 1. Age Compliance Check 👶
    # Consideration: ECDA maps enrollment age using the calendar year difference as of January 1st.
    calendar_age = admission_date.year - dob.year
    
    if calendar_age == 2:
        care_level = "Playgroup 🧸"
    elif calendar_age == 3:
        care_level = "Nursery 1 📚"
    elif calendar_age == 4:
        care_level = "Nursery 2 🎨"
    elif calendar_age == 5:
        care_level = "Kindergarten 1 🎒"
    elif calendar_age == 6:
        care_level = "Kindergarten 2 🎓"
    else:
        return {"error": "Age outside standard preschool brackets ⚠️"}

    # 2. Additional Subsidy Tier Evaluation (GHI) 📊
    # Consideration: Sequentially checks Gross Monthly Household Income against ECDA brackets.
    if ghi <= 3000:
        additional_subsidy = 400  # Lower-income tier maximum support
    elif ghi <= 6000:
        additional_subsidy = 250  # Mid-income tier support
    elif ghi <= 12000:
        additional_subsidy = 100  # Upper-middle tier support
    else:
        additional_subsidy = 0    # Above upper income ceiling

    # 3. Net Out-of-Pocket Fee Calculation 🧮
    # Consideration: Subtracts universal basic subsidies and income-tiered grants from the graph base fee, 
    # using max(0, ...) to ensure the net fee never drops below zero.
    net_fee = max(0, base_fee - basic_subsidy - additional_subsidy)
    
    return {
        "eligible_level": care_level,
        "additional_subsidy": additional_subsidy,
        "net_monthly_fee": net_fee
    }
```

Defines `evaluate_preschool_eligibility()` which:
- Determines the child's calendar age (year-based) and maps it to a care level.
- Evaluates additional subsidy tiers based on Gross Household Income (GHI).
- Computes the final `net_monthly_fee` as `max(0, base_fee - basic_subsidy - additional_subsidy)`.

Summary: This function centralizes eligibility logic and returns a small dictionary with `eligible_level`, `additional_subsidy`, and `net_monthly_fee`.
